# The ch01 teaser — one human scanpath on one image

Reproduces `results/foveation_mit1003/intro_scanpath_stim0091.png`, the single
figure in the thesis introduction.

Deliberately the simplest figure in the repo: one human's fixation sequence over
one stimulus, no model output at all. The point ch01 makes at that stage is
about eye movements — that viewing is a sequence of discrete fixations joined by
saccades, not a uniform sweep — not yet about DeepGaze III. The same stimulus is
reused by the ch03 mechanism figures and the ch04 demo so the thesis carries one
running example.

> No model, no GPU. Runs in seconds.

```python
import runpy
import sys
from pathlib import Path

REPO = Path.cwd()
while not (REPO / 'pyproject.toml').exists() and REPO != REPO.parent:
    REPO = REPO.parent

from IPython.display import Image, display


def run_script(name, *args):
    """Run scripts/<name> exactly as the command line would.

    These notebooks drive the same code the committed figure came from rather
    than reimplementing the plotting, so a notebook cannot silently disagree
    with what is in results/ and in the thesis.
    """
    sys.argv = [name, *[str(a) for a in args]]
    runpy.run_path(str(REPO / 'scripts' / name), run_name='__main__')


def show(*paths, width=1000):
    for p in paths:
        display(Image(filename=str(REPO / p), width=width))
```

## PARAMETERS — edit these

```python
STIM_IDX    = 91     # MIT1003 stimulus index
SUBJECT_IDX = None   # None = first subject with a usable scanpath on this image
```

```python
args = ['--stim-idx', STIM_IDX]
if SUBJECT_IDX is not None:
    args += ['--subject-idx', SUBJECT_IDX]
run_script('demo_intro_scanpath.py', *args)
```

```python
show(f'results/foveation_mit1003/intro_scanpath_stim{STIM_IDX:04d}.png', width=800)
```
